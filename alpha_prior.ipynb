```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("input_data/100Z_ld.csv")

df
```

```python
df_sel = df.iloc[1:4].copy()

df_sel
```

```python
df_sel['left']= (-1/df_sel['energy'])*np.log((df_sel['ld']-df_sel['errors'])/df_sel['ld'])
df_sel['right']= (-1/df_sel['energy'])*np.log((df_sel['ld']+df_sel['errors'])/df_sel['ld'])


df_sel['average']= (df_sel['left']+np.abs(df_sel['right']))/2
df_sel
```

```python
aveage_sigma = result = np.sqrt(1/(1/df_sel['average'] ** 2).sum())

print('aveage_sigma',aveage_sigma)
print('aveage_sigma^2',aveage_sigma**2)
```
