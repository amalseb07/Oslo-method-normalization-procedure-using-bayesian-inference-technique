```python
import numpy as np

import matplotlib.pyplot as plt
import scipy
import pandas as pd
```

```python
# load gsf parameter set 
gsf = np.loadtxt('output_data/gsf_post.txt')
print(gsf.shape)

```

```python
# making gsf the same size as NLD

gsf_sliced = gsf[:,0:3]

final_gsf = np.repeat(gsf_sliced, repeats=100, axis=0)

print(final_gsf.shape)
```

```python
#load NLD parameter set
ld = np.load("output_data/nld_post.npy")

print(ld.shape)
```

```python
# sampling randomly to reduce data set

rng = np.random.default_rng(45)

idx = rng.choice(10000, size=500, replace=False)

gsf_rand = final_gsf[idx]
ld_rand = ld[idx]

print(gsf_rand.shape)
print(ld_rand.shape)

```

```python
# store as file1 = NLD and file2 = GSF

np.savetxt('file1.txt',ld_rand,fmt ='%.6f')
np.savetxt('file2.txt',gsf_rand,fmt ='%.6f')
```
