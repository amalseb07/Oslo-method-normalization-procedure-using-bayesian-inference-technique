```python
import pandas as pd
import numpy as np
```

```python
dfs_p = {}
dfs_n = {}

for ptable in np.arange(0.05, 0.46, 0.05):
    ptable = round(ptable, 2)

    df_p = pd.read_table(
        f"../ld_positive_{ptable}.txt",
        delimiter=r"\s+"
    )

    df_n = pd.read_table(
        f"../ld_negative_{ptable}.txt",
        delimiter=r"\s+"
    )


    df_p['Ex'] = df_p['Ex'] - ptable
    df_n['Ex'] = df_n['Ex']- ptable

    if ptable >= 0.25:
        df_p = df_p[df_p['Ex'] >= 0]
        df_n = df_n[df_n['Ex'] >= 0]
    
    dfs_p[ptable] = df_p
    dfs_n[ptable] = df_n
    
```

```python
df_all_p = pd.concat(dfs_p.values(), ignore_index=True)

df_all_p["Ex"] = df_all_p["Ex"].round(2)
df_all_p = (
    df_all_p
    .sort_values(by="Ex", ascending=True)
    .reset_index(drop=True)
)

df_all_p = df_all_p.drop_duplicates(subset="Ex", keep="first")

df_all_p.to_pickle("master_base_ld_p.pkl")
```

```python
df_all_n = pd.concat(dfs_n.values(), ignore_index=True)


df_all_n["Ex"] = df_all_n["Ex"].round(2)
df_all_n = (
    df_all_n
    .sort_values(by="Ex", ascending=True)
    .reset_index(drop=True)
)

df_all_n = df_all_n.drop_duplicates(subset="Ex", keep="first")

df_all_n.to_pickle("master_base_ld_n.pkl")
```
